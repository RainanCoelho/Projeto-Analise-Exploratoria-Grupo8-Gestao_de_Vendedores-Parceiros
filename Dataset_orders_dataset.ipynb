PARTE 3: CORREÇÃO DE TIPOS





```python
colunas_datas = ["order_purchase_timestamp","order_approved_at","order_delivered_carrier_date","order_delivered_customer_date","order_estimated_delivery_date"]

def converter_datas(df,colunas):
  for coluna in colunas:
    df[coluna] = pd.to_datetime(df[coluna], errors="coerce")
  return df

df_osd = converter_datas(df_osd, colunas_datas)
df_osd.dtypes
```

**Explicação**: As colunas que possuem dados de datas, estão sendo lidas como "Object", mas o correto seria uma leitura das mesmas para "date-time", por isso a necessidade da alteração do tipo de cada uma

PARTE 6: PERÍODO DAS DATAS

```python
#Verificar se possue alguma data incoerente com a data dita no kaggle, que é de 2016 até 2018
for coluna in colunas_datas:
  abaixo_2016 = (df_osd[coluna].dt.year < 2016).sum()
  acima_2018 = (df_osd[coluna].dt.year > 2018).sum()
  print(f"{coluna}: abaixo de 2016 = {abaixo_2016} | acima de 2018: {acima_2018}")
```

**Explicação**: Não existe nenhum valor incoerente em relação a datações, todas estão dentro do período da busca dos dados

PARTE 7: COERÊNCIA DAS DATAS

```python
#Esses comandos dão nomes as colunas para facilitar na compreensão e na visualização de uma ordem de entrega de produtos:
#compra → aprovação → despacho → entrega ao cliente
compra = df_osd["order_purchase_timestamp"]
aprovacao = df_osd["order_approved_at"]
despacho = df_osd["order_delivered_carrier_date"]
entrega = df_osd["order_delivered_customer_date"]

print("Aprovação antes da compra:", (aprovacao < compra).sum()) #Verifica se tem alguma compra depois da aprovação
print("Despacho antes da compra:", (despacho < compra).sum()) #Verifica se tem alguma compra depois do despacho
print("Despacho antes da aprovação:", (despacho < aprovacao).sum()) #Verifica se tem aguma aprovação depois do despacho
print("Entrega antes do despacho:", (entrega < despacho).sum()) #Verifica se tem algum despacho depois da entrega
```

PARTE 8: TRATAMENTO DE INCONSISTÊNCIAS

```python
def anular_despacho_incoerente(df):
  df = df.copy()
  erro = df["order_delivered_carrier_date"] < df["order_purchase_timestamp"]
  df.loc[erro, "order_delivered_carrier_date"] = pd.NaT
  return df

print("Antes:", (df_osd["order_delivered_carrier_date"] < df_osd["order_purchase_timestamp"]).sum())
df_osd = anular_despacho_incoerente(df_osd)
print("Depois:", (df_osd["order_delivered_carrier_date"] < df_osd["order_purchase_timestamp"]).sum())
```

**Explicação**: Foram encontrados 166 pedidos com a data de despacho anterior a data da compra, o que é algo impossível de acontecer, e com isso, há uma distorção na análise, e faz com que o vendedor pareça adiantado. A data de despacho acabou que foi anulada, mas os pedidos foram mantidos na tabela, já que as demais informações são importantes para a tabela, já a inconsistência de despacho antes da aprovação e a de entregue antes do despacho foram mantidas, já que ambas não respondem as perguntas do cliente
