PARTE 8: TRATAMENTO DE INCONSISTÊNCIAS

```python
def anular_despacho_incoerente(df):
  df = df.copy()
  erro = df["order_delivered_carrier_date"] < df["order_purchase_timestamp"]
  df.loc[erro, "order_delivered_carrier_date"] = pd.NaT
  return df

print("Antes:", (df_osd["order_delivered_carrier_date"] < df_osd["order_purchase_timestamp"]).sum())
df_osd = anular_despacho_incoerente(df_osd)
print("Depois:", (df_osd["order_delivered_carrier_date"] < df_osd["order_purchase_timestamp"]).sum())
```

**Explicação**: Foram encontrados 166 pedidos com a data de despacho anterior a data da compra, o que é algo impossível de acontecer, e com isso, há uma distorção na análise, e faz com que o vendedor pareça adiantado. A data de despacho acabou que foi anulada, mas os pedidos foram mantidos na tabela, já que as demais informações são importantes para a tabela, já a inconsistência de despacho antes da aprovação e a de entregue antes do despacho foram mantidas, já que ambas não respondem as perguntas do cliente

PARTE 9: CRIAÇÃO DE NOVOS ATRIBUTOS (FEATURE ENGINEERING)

```python
def criar_features_orders(df):
    out = df.copy()

    # Diferença entre aprovação e compra convertida de segundos para horas
    out["tempo_aprovacao_horas"] = (
        out["order_approved_at"] - out["order_purchase_timestamp"]
    ).dt.total_seconds() / 3600

    # Diferença entre despacho (transportadora) e aprovação em horas
    tempo_despacho = (
        out["order_delivered_carrier_date"] - out["order_approved_at"]
    ).dt.total_seconds() / 3600
    # Substitui valores negativos por NaN mantendo os não-negativos
    out["tempo_despacho_horas"] = tempo_despacho.where(tempo_despacho >= 0, np.nan)

    # Diferença entre entrega ao cliente e despacho em dias (86400s = 24h)
    tempo_transporte = (
        out["order_delivered_customer_date"] - out["order_delivered_carrier_date"]
    ).dt.total_seconds() / 86400
    # Substitui valores negativos por NaN quando a entrega ocorre antes do despacho
    out["tempo_transporte_dias"] = tempo_transporte.where(tempo_transporte >= 0, np.nan)

    # Diferença entre entrega real e entrega estimada em dias (positivo = atraso)
    out["atraso_entrega_dias"] = (
        out["order_delivered_customer_date"] - out["order_estimated_delivery_date"]
    ).dt.total_seconds() / 86400

    # Flag binária (1/0) indicando se o tempo de despacho é válido (não nulo)
    out["flag_despacho_valido"] = out["tempo_despacho_horas"].notna().astype(int)

    # Flag de atraso quando ambas as datas existem; mantém NaN se faltar alguma
    out["flag_entregue_com_atraso"] = np.where(
        out["order_delivered_customer_date"].notna() & out["order_estimated_delivery_date"].notna(),
        (out["order_delivered_customer_date"] > out["order_estimated_delivery_date"]).astype(int),
        np.nan,
    )

    return out


novas_cols = [
    "tempo_aprovacao_horas",
    "tempo_despacho_horas",
    "tempo_transporte_dias",
    "atraso_entrega_dias",
    "flag_despacho_valido",
    "flag_entregue_com_atraso",
]

# Gera o novo DataFrame com as features criadas
df_osd_f2 = criar_features_orders(df_osd)

# Exibe estatísticas descritivas e contagem de nulos das novas colunas
print(df_osd_f2[novas_cols].describe())
print(df_osd_f2[novas_cols].isna().sum())
```

**EXPLICAÇÃO** : A partir das datas tratadas, foram criadas seis colunas para separar o tempo de cada etapa do pedido, isolando a demora do despacho do vendedor do tempo de aprovação e do transporte. Foi visto que a aprovação leva em média 10,42 horas, o despacho pelo vendedor leva 68,62 horas e o transporte leva 9,34 dias. Os tempos negativos de despacho e transporte viraram NaN para não distorcer as médias, cerca de 97% dos pedidos têm despacho válido e cerca de 8% dos entregues tiveram atraso.

PARTE 10: CODIFICAÇÃO DE DADOS CATEGÓRICOS

```python
def codificar_status_orders(df):
    out = df.copy()

    # Flag binária (1/0) indicando se o pedido foi entregue
    out["flag_pedido_entregue"] = (out["order_status"] == "delivered").astype(int)

    # Flag binária (1/0) indicando se o pedido foi cancelado ou ficou indisponível
    out["flag_pedido_cancelado"] = out["order_status"].isin(["canceled", "unavailable"]).astype(int)

    # Aplica One-Hot Encoding em order_status mantendo a coluna original e sem alterar IDs
    dummies = pd.get_dummies(out["order_status"], prefix="status", dtype=int)
    out = pd.concat([out, dummies], axis=1)

    return out


# Colunas originais para identificar apenas as novas colunas criadas
cols_originais = set(df_osd_f2.columns)

# Aplica a função e atualiza df_osd_f2
df_osd_f2 = codificar_status_orders(df_osd_f2)

# Lista das novas colunas geradas
novas_cols = [col for col in df_osd_f2.columns if col not in cols_originais]

# Exibe os totais (soma) de cada nova coluna criada
print("Totais por nova coluna:")
print(df_osd_f2[novas_cols].sum())

# Exibe as novas colunas e as primeiras linhas do DataFrame atualizado
print("Novas colunas:", novas_cols)
print(df_osd_f2.head())

# Exibe o total de pedidos, contagem das flags, quantidade de novas colunas e distribuição de order_status
print("Total de pedidos:", len(df_osd_f2))
print("Total flag_pedido_entregue:", df_osd_f2["flag_pedido_entregue"].sum())
print("Total flag_pedido_cancelado:", df_osd_f2["flag_pedido_cancelado"].sum())
print("Quantidade de novas colunas criadas:", len(novas_cols))
print("Distribuição de order_status:")
print(df_osd_f2["order_status"].value_counts(dropna=False))
```

**EXPLICAÇÃO** : O texto da coluna order_status foi transformado em dez colunas numéricas de 0 e 1, mantendo order_id e customer_id intactos para cruzar com os dados de vendedores e avaliações depois, e preservando order_status para filtros. Foi visto que dos 99.441 pedidos, 96.478 foram entregues e 1.234 estão cancelados ou indisponíveis, permitindo separar pedidos concluídos nas análises.

PARTE 11: NORMALIZAÇÃO E PADRONIZAÇÃO

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler


def escalonar_orders(df, colunas):
    out = df.copy()

    # Garante que as colunas sejam numéricas e converte valores infinitos (inf/-inf) em NaN
    dados_num = out[colunas].apply(pd.to_numeric, errors="coerce").replace([np.inf, -np.inf], np.nan)

    # Inicializa os escalonadores do scikit-learn
    minmax_scaler = MinMaxScaler()
    standard_scaler = StandardScaler()

    # Nomes das novas colunas com os sufixos solicitados
    cols_minmax = [f"{col}_minmax" for col in colunas]
    cols_zscore = [f"{col}_zscore" for col in colunas]

    # Atribui coluna a coluna para evitar problemas de alinhamento/indexação no Pandas
    minmax_vals = minmax_scaler.fit_transform(dados_num)
    zscore_vals = standard_scaler.fit_transform(dados_num)

    for i, col in enumerate(colunas):
        out[f"{col}_minmax"] = minmax_vals[:, i]
        out[f"{col}_zscore"] = zscore_vals[:, i]

    return out


cols_numericas = [
    "tempo_aprovacao_horas",
    "tempo_despacho_horas",
    "tempo_transporte_dias",
    "atraso_entrega_dias",
]

# Aplica o escalonamento e salva em df_osd_f2
df_osd_f2 = escalonar_orders(df_osd_f2, cols_numericas)

# Lista de todas as colunas escalonadas geradas
cols_escalonadas = [f"{col}_minmax" for col in cols_numericas] + [
    f"{col}_zscore" for col in cols_numericas
]

# Exibe as estatísticas descritivas das colunas escalonadas
print(df_osd_f2[cols_escalonadas].describe())
```

**EXPLICAÇÃO**: As quatro colunas de tempo e atraso foram colocadas nas escalas MinMax, entre 0 e 1, e Z-score, com média 0 e desvio padrão 1, criando oito colunas novas. As originais em horas e dias foram mantidas para facilitar a definição de regras claras de acompanhamento dos vendedores, enquanto o Z-score ajuda a enxergar casos muito fora da média. A contagem de registros válidos não mudou, confirmando que os NaN foram mantidos.

PARTE 12: SELEÇÃO DE ATRIBUTOS

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_selection import VarianceThreshold

cols_6_num = [
    "tempo_aprovacao_horas",
    "tempo_despacho_horas",
    "tempo_transporte_dias",
    "atraso_entrega_dias",
    "flag_despacho_valido",
    "flag_entregue_com_atraso",
]

# 1. Heatmap de correlação das 6 numéricas
corr_6 = df_osd_f2[cols_6_num].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr_6, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlação - 6 Variáveis Numéricas")
plt.tight_layout()
plt.show()

# 2. VarianceThreshold(0.01) nas 6 numéricas + colunas status_* em cópia imputada pela mediana
cols_status = [c for c in df_osd_f2.columns if c.startswith("status_")]
cols_vt = cols_6_num + cols_status

df_vt_temp = df_osd_f2[cols_vt].copy()
df_vt_temp = df_vt_temp.fillna(df_vt_temp.median(numeric_only=True))

vt = VarianceThreshold(threshold=0.01)
vt.fit(df_vt_temp)

baixa_variancia = [col for col, keep in zip(cols_vt, vt.get_support()) if not keep]
print("Colunas com variância <= 0.01:", baixa_variancia)


# 3. Função para remover colunas apontadas pelo VarianceThreshold e status_delivered sem alterar o original
def selecionar_atributos_orders(df, cols_remover=None):
    out = df.copy()

    # Se não for passada lista explícita, remove todas as colunas de baixa variância + status_delivered
    if cols_remover is None:
        status_cols = [c for c in out.columns if c.startswith("status_")]
        redundantes = [c for c in ["status_delivered"] if c in status_cols]
        cols_remover = list(dict.fromkeys(baixa_variancia + redundantes))

    out = out.drop(columns=cols_remover, errors="ignore")
    return out


print("Shape antes:", df_osd_f2.shape)
df_osd_sel = selecionar_atributos_orders(df_osd_f2)
print("Shape depois:", df_osd_sel.shape)
print("Colunas finais:", list(df_osd_sel.columns))
```

**EXPLICAÇÃO**: O gráfico de calor mostrou a relação entre as seis colunas numéricas, confirmando que não há repetição entre elas e que o atraso final se liga ao transporte e ao despacho. O teste de variação identificou os seis status raros da base e, junto com status_delivered, essas sete colunas foram removidas, reduzindo a tabela de 32 para 25 colunas com todas as métricas principais prontas para a análise dos vendedores.
